```python
import numpy as np
import pandas as pd

df = pd.read_excel(
    r"C:\Users\priya\Downloads\online+retail\Online Retail.xlsx"
)

df.head()
```

```python
df.info()
```

```python
data = df.copy()

data.isna().sum()
```

```python
customer_data = data.dropna(subset=["CustomerID"]).copy()

print("Rows before:", len(data))
print("Rows with a customer ID:", len(customer_data))
print("Rows excluded:", len(data) - len(customer_data))
```

```python
duplicate_count = customer_data.duplicated().sum()

print("Identical repeated rows:", duplicate_count)
```

```python
duplicate_rows = customer_data[
    customer_data.duplicated(keep=False)
]

duplicate_rows.sort_values(
    ["InvoiceNo", "StockCode", "InvoiceDate"]
).head(10)
```

```python
customer_unique = customer_data.drop_duplicates().copy()

print("Rows before:", len(customer_data))
print("Rows after:", len(customer_unique))
print("Rows removed:", len(customer_data) - len(customer_unique))
```

```python
cancelled = customer_unique["InvoiceNo"].astype(str).str.upper().str.startswith("C")

print("Cancellation rows:", cancelled.sum())
print("Negative quantities:", (customer_unique["Quantity"] < 0).sum())
print("Zero quantities:", (customer_unique["Quantity"] == 0).sum())
print("Zero or negative prices:", (customer_unique["UnitPrice"] <= 0).sum())
```

```python
purchase_mask = (
    ~cancelled
    & (customer_unique["Quantity"] > 0)
    & (customer_unique["UnitPrice"] > 0)
)

purchases = customer_unique.loc[purchase_mask].copy()
adjustments = customer_unique.loc[~purchase_mask].copy()

print("Purchase rows:", len(purchases))
print("Rows kept separately for review:", len(adjustments))
print("Total rows accounted for:", len(purchases) + len(adjustments))
```

```python
purchases["LineAmount"] = (
    purchases["Quantity"] * purchases["UnitPrice"]
)

purchases[
    ["InvoiceNo", "Quantity", "UnitPrice", "LineAmount"]
].head()
```

```python
print("First purchase:", purchases["InvoiceDate"].min())
print("Last purchase:", purchases["InvoiceDate"].max())
print("Unique customers:", purchases["CustomerID"].nunique())
```

```python
cutoff_date = pd.Timestamp("2011-09-01")
end_date = cutoff_date + pd.Timedelta(days=90)

history = purchases[
    purchases["InvoiceDate"] < cutoff_date
].copy()

future_90_days = purchases[
    (purchases["InvoiceDate"] >= cutoff_date)
    & (purchases["InvoiceDate"] < end_date)
].copy()

print("Historical purchase rows:", len(history))
print("Future purchase rows:", len(future_90_days))
print("90-day window ends at:", end_date)
```

```python
rfm = history.groupby("CustomerID").agg(
    LastPurchase=("InvoiceDate", "max"),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("LineAmount", "sum")
)

rfm["Recency"] = (
    cutoff_date - rfm["LastPurchase"].dt.normalize()
).dt.days

rfm = rfm[["Recency", "Frequency", "Monetary"]]

rfm.head()
```

```python
future_spending = future_90_days.groupby("CustomerID")["LineAmount"].sum()

customer_summary = rfm.copy()

customer_summary["ActualSpend90Days"] = (
    future_spending.reindex(customer_summary.index).fillna(0)
)

customer_summary.head()
```

```python
rfm.describe().round(2)
```

```python
rfm_log = np.log1p(rfm)

rfm_log.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = pd.DataFrame(
    scaler.fit_transform(rfm_log),
    index=rfm_log.index,
    columns=rfm_log.columns
)

rfm_scaled.head()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 7):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(rfm_scaled)

    score = silhouette_score(
        rfm_scaled,
        labels,
        sample_size=1500,
        random_state=42
    )

    scores.append({"Clusters": k, "SilhouetteScore": score})

cluster_scores = pd.DataFrame(scores)
cluster_scores
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

segments = rfm.copy()
segments["Cluster"] = kmeans.fit_predict(rfm_scaled)

cluster_profile = segments.groupby("Cluster").agg(
    Customers=("Recency", "size"),
    MedianRecency=("Recency", "median"),
    MedianFrequency=("Frequency", "median"),
    MedianSpending=("Monetary", "median"),
    TotalSpending=("Monetary", "sum")
)

cluster_profile["SpendingSharePct"] = (
    cluster_profile["TotalSpending"]
    / cluster_profile["TotalSpending"].sum()
    * 100
)

cluster_profile.round(2)
```

```python
segment_names = {
    0: "Active higher-value",
    1: "Less recent lower-frequency"
}

segments["Segment"] = segments["Cluster"].map(segment_names)

segments.head()
```

```python
segment_names = {
    0: "Higher-value / more frequent group",
    1: "Lower-value / less frequent group"
}

segments["Segment"] = segments["Cluster"].map(segment_names)
```

```python
train_cutoff = pd.Timestamp("2011-06-01")
train_end = train_cutoff + pd.Timedelta(days=90)

train_history = purchases[
    purchases["InvoiceDate"] < train_cutoff
].copy()

train_future = purchases[
    (purchases["InvoiceDate"] >= train_cutoff)
    & (purchases["InvoiceDate"] < train_end)
].copy()

print("Training historical rows:", len(train_history))
print("Training future rows:", len(train_future))
print("Training target window ends at:", train_end)
```

```python
train_customers = train_history.groupby("CustomerID").agg(
    LastPurchase=("InvoiceDate", "max"),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("LineAmount", "sum")
)

train_customers["Recency"] = (
    train_cutoff - train_customers["LastPurchase"].dt.normalize()
).dt.days

train_customers = train_customers[
    ["Recency", "Frequency", "Monetary"]
].copy()

train_spending = train_future.groupby("CustomerID")["LineAmount"].sum()

train_customers["ActualSpend90Days"] = (
    train_spending.reindex(train_customers.index).fillna(0)
)

train_customers.head()
```

```python
recent_start = cutoff_date - pd.Timedelta(days=90)

recent_purchases = history[
    history["InvoiceDate"] >= recent_start
]

recent_spending = recent_purchases.groupby("CustomerID")["LineAmount"].sum()

customer_summary["BaselinePrediction"] = (
    recent_spending.reindex(customer_summary.index).fillna(0)
)

from sklearn.metrics import mean_absolute_error

baseline_mae = mean_absolute_error(
    customer_summary["ActualSpend90Days"],
    customer_summary["BaselinePrediction"]
)

print(f"Baseline MAE: £{baseline_mae:,.2f}")
```

```python
%pip install xgboost
```

```python
from xgboost import XGBRegressor

features = ["Recency", "Frequency", "Monetary"]

X_train = train_customers[features]
y_train = train_customers["ActualSpend90Days"]

X_test = customer_summary[features]
y_test = customer_summary["ActualSpend90Days"]

print("Training customers:", len(X_train))
print("Evaluation customers:", len(X_test))
```

```python
model = XGBRegressor(
    objective="reg:squarederror",
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=1.0,
    random_state=42,
    n_jobs=2
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)

# Purchase spending cannot be negative
predictions = np.maximum(predictions, 0)

model_mae = mean_absolute_error(y_test, predictions)

print(f"Baseline MAE: £{baseline_mae:,.2f}")
print(f"XGBoost MAE: £{model_mae:,.2f}")

improvement = (baseline_mae - model_mae) / baseline_mae * 100
print(f"MAE improvement over baseline: {improvement:.2f}%")
```

```python
# Spending during the 90 days before the June cutoff
train_recent = train_history[
    train_history["InvoiceDate"] >= (
        train_cutoff - pd.Timedelta(days=90)
    )
]

train_recent_spending = train_recent.groupby(
    "CustomerID"
)["LineAmount"].sum()

train_customers["SpendLast90Days"] = (
    train_recent_spending.reindex(train_customers.index).fillna(0)
)

# We already calculated this for the September baseline
customer_summary["SpendLast90Days"] = (
    customer_summary["BaselinePrediction"]
)

train_customers[
    ["Recency", "Frequency", "Monetary", "SpendLast90Days"]
].head()
```

```python
features = ["Recency", "Frequency", "Monetary", "SpendLast90Days"]

X_train = train_customers[features]
X_test = customer_summary[features]

model.fit(X_train, y_train)

predictions = np.maximum(model.predict(X_test), 0)

model_mae = mean_absolute_error(y_test, predictions)
improvement = (baseline_mae - model_mae) / baseline_mae * 100

print(f"Baseline MAE: £{baseline_mae:,.2f}")
print(f"Updated XGBoost MAE: £{model_mae:,.2f}")
print(f"MAE improvement: {improvement:.2f}%")

# Keep the better-performing approach for this comparison
if model_mae < baseline_mae:
    selected_method = "XGBoost"
    selected_predictions = predictions
else:
    selected_method = "Previous 90-day spending baseline"
    selected_predictions = customer_summary["BaselinePrediction"].to_numpy()

customer_summary["PredictedSpend90Days"] = selected_predictions

print("Selected approach:", selected_method)
```

```python
results = customer_summary.copy()

results["Segment"] = segments["Segment"]

results["AbsoluteError"] = (
    results["ActualSpend90Days"] - results["PredictedSpend90Days"]
).abs()

print("Customers with the largest prediction errors:")

display(
    results.sort_values("AbsoluteError", ascending=False)[
        [
            "Recency",
            "Frequency",
            "Monetary",
            "Segment",
            "PredictedSpend90Days",
            "ActualSpend90Days",
            "AbsoluteError"
        ]
    ].head(10).round(2)
)
```

```python
business_summary = results.groupby("Segment").agg(
    Customers=("Recency", "size"),
    MedianRecency=("Recency", "median"),
    HistoricalSpending=("Monetary", "sum"),
    Predicted90DaySpending=("PredictedSpend90Days", "sum"),
    Actual90DaySpending=("ActualSpend90Days", "sum"),
    MAE=("AbsoluteError", "mean")
)

business_summary["HistoricalSpendingSharePct"] = (
    business_summary["HistoricalSpending"]
    / business_summary["HistoricalSpending"].sum()
    * 100
)

business_summary["SuggestedAction"] = [
    (
        "Prioritise service and retention; review inactive customers individually"
        if segment == "Higher-value / more frequent group"
        else "Test a targeted reactivation campaign against a control group"
    )
    for segment in business_summary.index
]

display(business_summary.round(2))

top10_error_share = (
    results["AbsoluteError"].nlargest(10).sum()
    / results["AbsoluteError"].sum()
    * 100
)

print(f"Top 10 customers' share of total absolute error: {top10_error_share:.1f}%")
```

```python
import matplotlib.pyplot as plt

labels = ["Higher-value group", "Lower-value group"]
colors = ["#2563eb", "#f59e0b"]

summary = business_summary.reindex([
    "Higher-value / more frequent group",
    "Lower-value / less frequent group"
])

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# 1. Number of customers
axes[0].bar(labels, summary["Customers"], color=colors)
axes[0].set_title("Customers by segment")
axes[0].set_ylabel("Number of customers")

# 2. Historical purchase spending share
axes[1].bar(
    labels,
    summary["HistoricalSpendingSharePct"],
    color=colors
)
axes[1].set_title("Historical spending share")
axes[1].set_ylabel("Share (%)")
axes[1].set_ylim(0, 100)

# 3. Predicted versus actual spending
x = np.arange(len(labels))
width = 0.35

axes[2].bar(
    x - width / 2,
    summary["Predicted90DaySpending"] / 1000,
    width,
    label="Baseline prediction",
    color="#94a3b8"
)

axes[2].bar(
    x + width / 2,
    summary["Actual90DaySpending"] / 1000,
    width,
    label="Actual",
    color="#2563eb"
)

axes[2].set_xticks(x)
axes[2].set_xticklabels(labels)
axes[2].set_title("90-day purchase spending")
axes[2].set_ylabel("Spending (£ thousands)")
axes[2].legend()

for ax in axes:
    ax.tick_params(axis="x", labelrotation=15)

fig.suptitle(
    "Customer segments at 1 September 2011 | Spending excludes refunds"
)
plt.tight_layout()
plt.show()
```

```python
from pathlib import Path
import json
import joblib

output_dir = Path("project_outputs")
output_dir.mkdir(exist_ok=True)

# Save tables, including customer IDs
results.to_csv(output_dir / "customer_results.csv", index=True)
business_summary.to_csv(output_dir / "segment_summary.csv", index=True)
cluster_scores.to_csv(output_dir / "cluster_scores.csv", index=False)

# Save the chart created in the previous cell
fig.savefig(
    output_dir / "customer_summary.png",
    dpi=200,
    bbox_inches="tight"
)

# Save clustering: apply log1p before using this scaler
joblib.dump(
    {
        "scaler": scaler,
        "kmeans": kmeans,
        "features": ["Recency", "Frequency", "Monetary"],
        "transformation": "numpy.log1p",
        "segment_names": segment_names
    },
    output_dir / "segmentation_model.joblib"
)

# Save XGBoost as the comparison model
model.save_model(str(output_dir / "xgboost_comparison.json"))

# Document the selected approach and evaluation limits
metadata = {
    "selected_method": selected_method,
    "baseline_rule": "Predict spending equal to the previous 90 days",
    "baseline_mae_gbp": float(baseline_mae),
    "updated_xgboost_mae_gbp": float(model_mae),
    "prediction_features": features,
    "training_cutoff": str(train_cutoff),
    "evaluation_cutoff": str(cutoff_date),
    "evaluation_end_exclusive": str(end_date),
    "target": "90-day purchase spending excluding refunds",
    "limitations": [
        "September results were inspected during model development; "
        "this is an exploratory comparison.",
        "Only customers known before each cutoff are included.",
        "Exact duplicate rows were removed as a cleaning assumption.",
        "Returns are retained separately but excluded from spending.",
        "Segment actions are proposals, not measured campaign effects."
    ]
}

with open(output_dir / "project_metadata.json", "w", encoding="utf-8") as f:
    json.dump(metadata, f, indent=2)

print("Saved files to:", output_dir.resolve())
print("Selected prediction method:", selected_method)
```
